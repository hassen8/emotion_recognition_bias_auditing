# Load RAF-DB lables + build file paths

```python
def load_rafdb_df(image_dir, label_file, split="train"):
    df = pd.read_csv(label_file, sep=" ", header=None, names=["filename", "label"])

    if split == "train":
        df = df[df["filename"].str.startswith("train")]
    elif split == "test":
        df = df[df["filename"].str.startswith("test")]
    else:
        raise ValueError("split must be 'train' or 'test'.")

    df["aligned_filename"] = df["filename"].str.replace(".jpg", "_aligned.jpg", regex=False)
    df["file_path"] = df["aligned_filename"].apply(lambda x: os.path.join(image_dir, x))
    df.drop(columns=["aligned_filename"], inplace=True)

    # map RAF numeric to emotion string
    df["emotion_label"] = df["label"].map(RAF_ID_TO_EMO)

    df = df[df["file_path"].apply(os.path.exists)].reset_index(drop=True)

    # sanity
    missing = df["emotion_label"].isna().sum()
    if missing > 0:
        raise RuntimeError(f"{missing} labels could not be mapped. Check label file format.")

    return df

df = load_rafdb_df(IMAGE_DIR, LABEL_FILE, split="train")
print(df.head())
print("N:", len(df))
print(df["emotion_label"].value_counts())
```

```python
import numpy as np
import json

def make_json_safe(obj):
    if isinstance(obj, dict):
        return {k: make_json_safe(v) for k, v in obj.items()}
    if isinstance(obj, list):
        return [make_json_safe(v) for v in obj]
    if isinstance(obj, tuple):
        return [make_json_safe(v) for v in obj]
    if isinstance(obj, (np.floating,)):
        return float(obj)
    if isinstance(obj, (np.integer,)):
        return int(obj)
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    return obj
```

```python
def deepface_analyze_emotion(path):
    out = DeepFace.analyze(
        img_path=path,
        actions=["emotion"],
        enforce_detection=False
    )
    return out[0]

def extract_proba_from_deepface_entry(entry, emotions=EMOTIONS):
    # DeepFace returns 0..100
    emo_dict = entry["emotion"]
    probs = np.array([emo_dict[e] for e in emotions], dtype=np.float32) / 100.0
    # normalize small drift
    s = probs.sum()
    if s > 0:
        probs = probs / s
    return probs

# Run
paths = df["file_path"].tolist()

results = []
y_pred_proba = np.zeros((len(paths), len(EMOTIONS)), dtype=np.float32)

for i, p in enumerate(tqdm(paths)):
    entry = deepface_analyze_emotion(p)
    results.append(entry)
    y_pred_proba[i] = extract_proba_from_deepface_entry(entry, EMOTIONS)

OUT_JSON = "/content/drive/MyDrive/deepface_results_raf_train.json"

serializable_results = make_json_safe(results)

with open(OUT_JSON, "w") as f:
    json.dump(serializable_results, f, indent=2)

print("Saved:", OUT_JSON)
y_pred = np.argmax(y_pred_proba, axis=1)
```

# Calibration: ECE + Reliability diagram

```python
def compute_ece(y_true, y_prob, n_bins=15):
    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)
    accuracies = (predictions == y_true).astype(np.float32)

    bin_edges = np.linspace(0, 1, n_bins + 1)
    ece = 0.0

    for i in range(n_bins):
        lo, hi = bin_edges[i], bin_edges[i+1]
        in_bin = (confidences > lo) & (confidences <= hi)
        prop = in_bin.mean()

        if prop > 0:
            acc_bin = accuracies[in_bin].mean()
            conf_bin = confidences[in_bin].mean()
            ece += abs(acc_bin - conf_bin) * prop

    return float(ece)

ece = compute_ece(y_true, y_pred_proba, n_bins=15)
print(f"ECE: {ece:.4f}")
```

```python
def plot_reliability_diagram(y_true, y_prob, n_bins=15, title="Reliability Diagram"):
    confidences = np.max(y_prob, axis=1)
    predictions = np.argmax(y_prob, axis=1)
    correct = (predictions == y_true)

    bin_edges = np.linspace(0, 1, n_bins + 1)
    bin_acc, bin_conf, bin_count = [], [], []

    for i in range(n_bins):
        lo, hi = bin_edges[i], bin_edges[i+1]
        in_bin = (confidences > lo) & (confidences <= hi)
        c = in_bin.sum()
        bin_count.append(c)

        if c > 0:
            bin_acc.append(correct[in_bin].mean())
            bin_conf.append(confidences[in_bin].mean())
        else:
            bin_acc.append(0.0)
            bin_conf.append((lo + hi)/2)

    plt.figure(figsize=(8, 8))
    plt.plot([0, 1], [0, 1], "k:", label="Perfect calibration")
    plt.plot(bin_conf, bin_acc, "s-", label="Model")
    plt.ylabel("Observed accuracy")
    plt.xlabel("Predicted confidence")
    plt.title(title)
    plt.ylim(0, 1); plt.xlim(0, 1)
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()

    plt.figure(figsize=(8, 3))
    plt.hist(confidences, bins=bin_edges, edgecolor="black", alpha=0.7)
    plt.xlabel("Confidence")
    plt.ylabel("Count")
    plt.title("Confidence Histogram")
    plt.grid(True, alpha=0.3)
    plt.show()

plot_reliability_diagram(y_true, y_pred_proba, n_bins=15)
```

```python
import tensorflow as tf
from tensorflow.keras.layers import Conv2D, DepthwiseConv2D, SeparableConv2D, Dense

def build_gradcam_model_from_sequential(base_model, input_shape=(48, 48, 1)):
    """
    Deterministic Grad-CAM model builder for models ending with Dense(activation=softmax).

    Key property: conv feature maps and logits are produced from the SAME forward graph,
    so gradients are never None (unless something else is truly wrong).

    Returns:
      gradcam_model: outputs [target_activation, logits]
    """
    last = base_model.layers[-1]
    if not (isinstance(last, Dense) and getattr(last, "activation", None) == tf.keras.activations.softmax):
        raise ValueError(
            "Expected last layer to be Dense(activation=softmax). "
            f"Got: {type(last)} activation={getattr(last, 'activation', None)}"
        )

    gc_input = tf.keras.Input(shape=input_shape)
    x = gc_input

    target_activation = None
    target_layer_name = None

    # Run only up to (excluding) the final Dense-softmax, tracking last conv activation
    for layer in base_model.layers[:-1]:
        x = layer(x)
        if isinstance(layer, (Conv2D, DepthwiseConv2D, SeparableConv2D)):
            target_activation = x
            target_layer_name = layer.name

    if target_activation is None:
        raise ValueError("No conv-like layer found for Grad-CAM.")

    # Build logits from the same tensor stream(right before final Dense-softmax)
    dense_softmax = base_model.layers[-1]
    dense_logits = tf.keras.layers.Dense(
        units=dense_softmax.units,
        activation=None,
        name=dense_softmax.name + "_logits"
    )
    logits = dense_logits(x)
    dense_logits.set_weights(dense_softmax.get_weights())

    print(f"[Grad-CAM] Target conv: {target_layer_name} | Using logits (Dense softmax -> Dense linear).")

    return tf.keras.Model(inputs=gc_input, outputs=[target_activation, logits], name="gradcam_model_logits")
```

```python
gradcam_model = build_gradcam_model_from_sequential(base_model)

dummy = tf.zeros((1, 48, 48, 1), dtype=tf.float32)
fmaps, logits = gradcam_model(dummy, training=False)
print("Feature maps:", fmaps.shape, "Logits:", logits.shape)
```

## Preprocess RAF images for the emotion model

For GRAD-CAM we need to feed images in exactly the format the model expects


*   Grayscale faces
*   48x48 resolution

*   Normalized pixel values ([0,1])
*   Input shape (1,48,48,1) - (batch, height, width, channel)

We will therefore convert the RAF-DB aligned BGR images to this format before feeding them into the Grad-CAM pipeline.





```python
def preprocess_for_emotion(img_bgr):
    """
    Convert RAF aligned face image into a tensor of shape
     (1, 48, 48, 1) for the emotion model.
    """
    gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
    # Resize to 48x48
    resized = cv2.resize(gray, (48, 48), interpolation=cv2.INTER_AREA)
    # Normalize to [0,1]
    norm = resized.astype("float32") / 255.0
    # Add channel + batch dims: (1, 48, 48, 1)
   # tensor = norm.reshape(1, 48, 48, 1)
    return norm.reshape(1, 48, 48, 1)
```

# Grad-CAM heatmap function

### Grad-CAM on DeepFace Emotion Model

Given a CNN and a target class $c$ (e.g., "happy"), Grad-CAM operates on the **last
convolutional feature maps** $A^k$ (where $k$ indexes channels) and the class
score $y^c$ (the logit or pre-softmax value for class $c$).

Formally, let the preprocessed input be
$$
x \in \mathbb{R}^{1 \times 48 \times 48 \times 1}.
$$

1. **Forward pass**

   We run a forward pass through `gradcam_model` to obtain:

   - $A^k$: feature maps from the last convolutional layer
   - $y^c$: class scores for all emotion classes

2. **Select target class**

   We choose the target class $c$ either as:
   - the predicted class $c = \arg\max y$, or  
   - a manually specified class index.

3. **Gradients w.r.t. feature maps**

   Using automatic differentiation (`tf.GradientTape`), we compute the gradient
   of the target class score with respect to each feature map:

   $$
   \frac{\partial y^c}{\partial A^k_{ij}} \, .
   $$

4. **Channel importance weights**

   We spatially average these gradients over the feature map to obtain a single
   **importance weight** $\alpha_k^c$ per channel:

   $$
   \alpha_k^c = \frac{1}{Z} \sum_i \sum_j \frac{\partial y^c}{\partial A^k_{ij}}
   $$

   where $Z$ is the number of spatial locations (height $\times$ width).

5. **Grad-CAM map**

   We then form the Grad-CAM map as a weighted sum of the feature maps,
   followed by a ReLU to keep only positive contributions:

   $$
   L_{\text{Grad-CAM}}^c = \mathrm{ReLU}\left( \sum_k \alpha_k^c A^k \right).
   $$

6. **Normalization and upsampling**

   Finally, we normalize $L_{\text{Grad-CAM}}^c$ to the range $[0,1]$ and
   upsample it to the original image resolution to obtain a **heatmap**
   showing which pixels contribute most to class $c$.

In our project, we apply this procedure to a subset of RAF-DB samples to inspect how
DeepFace’s internal attention patterns relate to facial regions (eyes, mouth, etc.)
and to known differences in neurodivergent expression (e.g., reduced eye contact,
flatter affect).


```python
def make_gradcam_heatmap(img_tensor, class_index=None):
    img_tensor = tf.cast(img_tensor, tf.float32)

    with tf.GradientTape() as tape:
        conv_outputs, scores = gradcam_model(img_tensor, training=False)
        tape.watch(conv_outputs)

        if class_index is None:
            class_index = int(tf.argmax(scores[0]).numpy())
        else:
            class_index = int(class_index)

        class_score = scores[0, class_index]

    grads = tape.gradient(class_score, conv_outputs)
    if grads is None:
        raise RuntimeError("Gradients are None. Target layer might be disconnected.")

    pooled_grads = tf.reduce_mean(grads, axis=(1, 2))  # (1, C)
    conv_outputs = conv_outputs[0]                      # (H, W, C)
    pooled_grads = pooled_grads[0]                      # (C,)

    heatmap = tf.reduce_sum(conv_outputs * pooled_grads, axis=-1)
    heatmap = tf.nn.relu(heatmap)

    heatmap = heatmap - tf.reduce_min(heatmap)
    heatmap = heatmap / (tf.reduce_max(heatmap) + 1e-8)

    return heatmap.numpy(), scores.numpy()[0]
```

# Visualizing Grad-CAM on RAF images

We now apply Grad-CAM to RAF-DB samples. For each image:

1. We load the pre-aligned BGR face from RAF-DB.
2. We preprocess it to (1, 48, 48, 1) for the emotion model.
3. We compute the Grad-CAM heatmap.
4. We upsample the heatmap to the original image size and overlay it
   with a color map to highlight important regions.

This gives a qualitative view of which facial regions (eyes, mouth, nose etc.)
the model uses to classify different emotions.

### Overlay + example visualization


```python
def overlay_gradcam(face_rgb_uint8, cam_map_01, alpha=0.30):
    """
    face_rgb_uint8: (H,W,3) uint8 RGB  <-- IMPORTANT (RGB, not BGR)
    cam_map_01:     (h,w) float map (any range); we normalize to [0,1]

    Returns:
      cam_uint8: (H,W) uint8
      heat_rgb:  (H,W,3) uint8
      overlay:   (H,W,3) uint8
    """
    H, W = face_rgb_uint8.shape[:2]

    cam_resized = cv2.resize(cam_map_01.astype(np.float32), (W, H), interpolation=cv2.INTER_LINEAR)
    cam_resized = np.maximum(cam_resized, 0)

    mn, mx = float(cam_resized.min()), float(cam_resized.max())
    cam_norm = (cam_resized - mn) / (mx - mn + 1e-8)

    cam_uint8 = np.uint8(255 * cam_norm)

    heat_bgr = cv2.applyColorMap(cam_uint8, cv2.COLORMAP_JET)
    heat_rgb = cv2.cvtColor(heat_bgr, cv2.COLOR_BGR2RGB)

    overlay = cv2.addWeighted(face_rgb_uint8, 1 - alpha, heat_rgb, alpha, 0)
    return cam_uint8, heat_rgb, overlay
```

## Single example

```python
import numpy as np

def softmax_np(x):
    x = np.asarray(x, dtype=np.float32)
    x = x - np.max(x)
    ex = np.exp(x)
    return ex / (np.sum(ex) + 1e-8)

def pretty_pred(true_label, pred_label, scores, class_names, topk=3):
    scores = np.asarray(scores, dtype=np.float32)

    probs = softmax_np(scores)   # only for display
    order = np.argsort(probs)[::-1]

    top_lines = []
    for j in order[:topk]:
        top_lines.append(f"{class_names[j]:<9}  logit={scores[j]:>7.3f}   p={probs[j]:>6.3f}")

    print(f"GT   : {true_label}")
    print(f"PRED : {pred_label}   (p={probs[class_names.index(pred_label)]:.3f})")
    print("Top predictions:")
    for line in top_lines:
        print("  -", line)
```

## **Aggregated Grad-CAM Analysis (Beyond Single Images)**

### **Per-emotion Average Grad-CAM Heatmaps**

### Per-emotion Average Grad-CAM Heatmaps

Single Grad-CAM maps are useful but noisy. To capture **systematic patterns**, we compute an **average Grad-CAM heatmap per emotion**.

For each image in a subset:

1. We compute the Grad-CAM heatmap $H^{(n)}$ for the model's **predicted** class.
2. We group heatmaps by predicted emotion (e.g., "happy", "sad").
3. For each emotion \( e \), we compute the **mean heatmap**:

$$
\bar{H}_e = \frac{1}{N_e} \sum_{n \in \mathcal{I}_e} H^{(n)}
$$

where:
- $ \mathcal{I}_e $ is the set of indices whose predicted label is $e$,  
- $ N_e = |\mathcal{I}_e| $ is the number of samples in that set.  

TThe resulting $ \bar{H}_e $ shows the canonical attention pattern for emotion $e$ (for example, a mouth-centric pattern for "happy").

```python
from collections import defaultdict

N_SAMPLES = 500
CONF_TH = 0.6

rng = np.random.default_rng(0)
subset_indices = rng.choice(len(image_data), size=min(N_SAMPLES, len(image_data)), replace=False)

per_emotion_heatmaps = defaultdict(list)
counts_total = defaultdict(int)

for idx in tqdm(subset_indices):
    img_bgr = image_data[idx]
    img_tensor = preprocess_for_emotion(img_bgr)
    heatmap, scores = make_gradcam_heatmap(img_tensor)

    pred_idx = int(np.argmax(scores))
    pred_label = EMOTIONS[pred_idx]
    conf = float(np.max(scores))

    counts_total[pred_label] += 1
    if conf >= CONF_TH:
        per_emotion_heatmaps[pred_label].append(heatmap)

print("Total sampled per label:", dict(counts_total))
print("Kept (conf>=th):", {k: len(v) for k, v in per_emotion_heatmaps.items()})
```

```python
avg_heatmaps = {}
for label, hmaps in per_emotion_heatmaps.items():
    if len(hmaps) == 0:
        continue
    avg_heatmaps[label] = np.mean(np.stack(hmaps, axis=0), axis=0)

plt.figure(figsize=(14, 6))
for i, label in enumerate(EMOTIONS):
    if label not in avg_heatmaps:
        continue
    vis = cv2.resize(avg_heatmaps[label], (128, 128), interpolation=cv2.INTER_CUBIC)
    vis = (vis - vis.min()) / (vis.max() - vis.min() + 1e-8)

    plt.subplot(2, 4, i+1)
    plt.title(f"{label}\n(avg Grad-CAM)")
    plt.imshow(vis, cmap="jet")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

### **Region-level atention**

Grad-CAM heatmaps $H \in \mathbb{R}^{H \times W}$
 capture where the model looks,
but we need want **numeric** summaries.

We define approximate regions on the face (eyes, mouth, nose/center, other)
using fixed relative coordinates, assuming RAF-DB aligned faces.

Given a resized heatmap \( H \) over the original image resolution, let:

- $R_{\text{eyes}}$ = set of pixels in the eyes region  
- $R_{\text{mouth}}$ = set of pixels in the mouth region  
- $R_{\text{nose}}$ = set of pixels in a central band (nose/center)  
- $R_{\text{other}}$ = all remaining pixels  


We define the **attention fraction** for each region as:

$$
f_{\text{region}} = \frac{\sum_{(i,j) \in R_{\text{region}}} H_{ij}}
                         {\sum_{i,j} H_{ij}}
$$

These fractions satisfy:

$$
f_{\text{eyes}} + f_{\text{mouth}} + f_{\text{nose}} + f_{\text{other}} = 1
$$

By aggregating these fractions per emotion, we can say, for example:

- "On average, 65% of 'happy' attention is on the mouth region,"
- "For 'sad', attention shifts to the eye region,"
- etc.

This forms a quantitative link between model behavior and neurotypical facial
expression patterns (e.g., strong mouth-based happiness).


The following cell defines region definitions on the face.

```python
def _clip_box(x1, y1, x2, y2, W, H):
    x1 = int(np.clip(x1, 0, W - 1))
    x2 = int(np.clip(x2, 0, W))
    y1 = int(np.clip(y1, 0, H - 1))
    y2 = int(np.clip(y2, 0, H))
    if x2 <= x1: x2 = min(W, x1 + 1)
    if y2 <= y1: y2 = min(H, y1 + 1)
    return x1, y1, x2, y2
```

```python
def get_face_regions(img_bgr):
    H, W = img_bgr.shape[:2]

    eye_y1, eye_y2 = 0.18 * H, 0.42 * H
    eye_x1, eye_x2 = 0.18 * W, 0.82 * W

    nose_y1, nose_y2 = 0.35 * H, 0.65 * H
    nose_x1, nose_x2 = 0.35 * W, 0.65 * W

    mouth_y1, mouth_y2 = 0.62 * H, 0.88 * H
    mouth_x1, mouth_x2 = 0.25 * W, 0.75 * W

    return {
        "eyes":  _clip_box(eye_x1,  eye_y1,  eye_x2,  eye_y2,  W, H),
        "nose":  _clip_box(nose_x1, nose_y1, nose_x2, nose_y2, W, H),
        "mouth": _clip_box(mouth_x1, mouth_y1, mouth_x2, mouth_y2, W, H),
    }
```

This will compute region fractions for one sample.

```python
def compute_region_fractions(img_bgr, heatmap, eps=1e-8, priority=("eyes","nose","mouth")):
    """
      Args:
          img_bgr: (H,W,3) BGR image
          heatmap: 2D array (h,w) Grad-CAM heatmap in [0,1] (or >=0)
          eps: small constant
          priority: order used to resolve overlaps (first wins)

      Returns:
          fractions dict with keys: eyes, nose, mouth, other
          (sums to ~1.0)
    """
    H, W = img_bgr.shape[:2]
    hm = cv2.resize(heatmap, (W, H), interpolation=cv2.INTER_LINEAR)
    hm = np.maximum(hm.astype(np.float32), 0.0)

    total = float(hm.sum()) + eps
    regions = get_face_regions(img_bgr)

    assigned = np.zeros((H, W), dtype=bool)
    region_sums = {}

    for name in priority:
        x1, y1, x2, y2 = regions[name]
        mask = np.zeros((H, W), dtype=bool)
        mask[y1:y2, x1:x2] = True
        mask = mask & (~assigned)
        assigned |= mask
        region_sums[name] = float(hm[mask].sum())

    other_sum = float(hm[~assigned].sum())
    fracs = {k: v/total for k, v in region_sums.items()}
    fracs["other"] = other_sum/total

    s = sum(fracs.values())
    if s > 0:
        for k in fracs:
            fracs[k] /= s

    return fracs
```

# **Results analysis**

We used the DeepFace emotion model directly and attached a Grad-CAM pipeline to
its last convolutional layer. For each RAF-DB image, we fed the aligned face through the same preprocessing as DeepFace’s emotion pipeline, obtained the model’s predicted emotion, and then computed the gradient of that class score with respect to the feature maps of the last conv layer. We averaged these gradients to get channel-wise importance weights, we then combined them with the feature maps, and applied ReLU to obtain a class-specific importance map. Then after up sampling this to the input resolution and overlaying it on the original face, we obtained heatmaps that show which face regions the model relies on when classifying emotions.

We then compared these attention patters like for example the strong focus on
the mouth for "*happy*" and the heavy reliance on the eye region for some emotions, with literature on neurodivergent expression patters, such as reduced or minimal eue contact and flatter affect. this provides us with a concrete evidence that the model's internal logic is primarily aligned with neurotypical expressive normals which can be problematic, inducing bias to neurodireverse individuals.

# **Occlusion**

```python
import numpy as np
import cv2

def occlusion_sensitivity_map_48(img_tensor, class_index=None, patch=6, stride=3, baseline="mean"):
    """
    img_tensor: (1,48,48,1) float32 in [0,1]
    class_index: target class index; if None, uses model's argmax on logits
    patch: occlusion square size in 48x48 coordinates
    stride: step size
    baseline: "mean" or float value in [0,1]

    Returns:
      occ_map: (48,48) float32, larger = more important region
      class_index: int used target
      base_logits: (num_classes,) float32
    """
    x = img_tensor.astype(np.float32).copy()

    # baseline logits
    _, base_scores = gradcam_model(x, training=False)   # base_scores are logits
    base_scores = base_scores.numpy()[0].astype(np.float32)

    if class_index is None:
        class_index = int(np.argmax(base_scores))

    base_target = float(base_scores[class_index])

    if baseline == "mean":
        fill = float(np.mean(x))
    else:
        fill = float(baseline)

    H, W = 48, 48
    occ_accum = np.zeros((H, W), dtype=np.float32)
    occ_count = np.zeros((H, W), dtype=np.float32)

    # slide patch
    for y in range(0, H - patch + 1, stride):
        for x0 in range(0, W - patch + 1, stride):
            x_occ = x.copy()
            x_occ[:, y:y+patch, x0:x0+patch, :] = fill

            _, scores_occ = gradcam_model(x_occ, training=False)
            scores_occ = scores_occ.numpy()[0].astype(np.float32)
            occ_target = float(scores_occ[class_index])

            drop = base_target - occ_target  # positive drop means region mattered

            occ_accum[y:y+patch, x0:x0+patch] += drop
            occ_count[y:y+patch, x0:x0+patch] += 1.0

    occ_map = occ_accum / (occ_count + 1e-8)

    # Keep only positive influence; normalize to [0,1]
    occ_map = np.maximum(occ_map, 0)
    mn, mx = float(occ_map.min()), float(occ_map.max())
    occ_map_norm = (occ_map - mn) / (mx - mn + 1e-8)

    return occ_map_norm.astype(np.float32), class_index, base_scores
```

# Occlusion overlay

```python
def occlusion_on_face(img_bgr, patch=6, stride=3, alpha=0.45):
    """
    img_bgr: (H,W,3) uint8
    Returns:
      pred_label (str)
      conf_softmax (float) for display
      occ_map_48 (48,48) float32 [0,1]
      occ_uint8 (H,W) uint8
      overlay_rgb (H,W,3) uint8
    """
    img_tensor = preprocess_for_emotion(img_bgr)  # (1,48,48,1) float [0,1]

    occ_map_48, class_index, logits = occlusion_sensitivity_map_48(
        img_tensor, class_index=None, patch=patch, stride=stride, baseline="mean"
    )

    pred_idx = int(np.argmax(logits))
    pred_label = EMOTIONS[pred_idx]

    # softmax for human readable confidence
    probs = np.exp(logits - np.max(logits))
    probs = probs / (np.sum(probs) + 1e-8)
    conf = float(probs[pred_idx])

    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

    occ_uint8, _, overlay_rgb = overlay_gradcam(img_rgb, occ_map_48, alpha=alpha)

    return pred_label, conf, occ_map_48, occ_uint8, overlay_rgb
```

```python
from lime import lime_image
from skimage.segmentation import slic
from skimage.util import img_as_ubyte
```

```python
def preprocess_for_emotion_from_rgb(img_rgb):
    """
    img_rgb: (H,W,3) RGB uint8/float
    returns: (48,48,1) float32 in [0,1]
    """
    if img_rgb.dtype != np.uint8:
        # Lime may pass float images
        img_rgb = np.clip(img_rgb, 0, 255).astype(np.uint8)

    gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
    resized = cv2.resize(gray, (48, 48), interpolation=cv2.INTER_AREA)
    x = resized.astype(np.float32) / 255.0
    return x[..., None]  # (48,48,1)
```

```python
def deepface_predict_proba_lime(images):
    """
    LIME calls this with a batch of perturbed images.
    images: list/array of (H,W,3) RGB (uint8 or float)
    returns: (N,7) float32 probabilities aligned with EMOTIONS
    """

    X = np.stack([preprocess_for_emotion_from_rgb(im) for im in images], axis=0)
    # base_model ends with softmax = > probabilities
    probs = base_model.predict(X, verbose=0).astype(np.float32)

    # Safety normalize
    s = probs.sum(axis=1, keepdims=True)
    probs = probs / (s + 1e-8)
    return probs
```

```python
LIME_OUT = "/content/lime_outputs"
os.makedirs(LIME_OUT, exist_ok=True)

def lime_on_face(
    idx,
    n_segments=120,
    compactness=10,
    sigma=1,
    num_samples=1500,
    num_features=10,
    positive_only=True,
    hide_color=0,
    alpha_dim=0.45,
    save=True
):
    idx = int(idx)
    img_bgr = image_data[idx]
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    true_label = df.loc[idx, "emotion_label"]

    probs0 = deepface_predict_proba_lime([img_rgb])[0]
    pred_idx = int(np.argmax(probs0))
    pred_label = EMOTIONS[pred_idx]
    conf = float(probs0[pred_idx])

    segments = slic(
        img_as_ubyte(img_rgb),
        n_segments=int(n_segments),
        compactness=float(compactness),
        sigma=float(sigma),
        start_label=0,
    )

    explainer = lime_image.LimeImageExplainer(random_state=0)

    explanation = explainer.explain_instance(
        img_rgb,
        classifier_fn=deepface_predict_proba_lime,
        top_labels=1,
        hide_color=int(hide_color),
        num_samples=int(num_samples),
        segmentation_fn=lambda _: segments,
    )

    cutout, mask = explanation.get_image_and_mask(
        label=pred_idx,
        positive_only=bool(positive_only),
        num_features=int(num_features),
        hide_rest=True,
    )

    cutout_uint8 = np.clip(cutout, 0, 255).astype(np.uint8)

    # overlay
    overlay = img_rgb.copy().astype(np.float32)
    overlay[mask == 0] *= (1.0 - float(alpha_dim))
    overlay = np.clip(overlay, 0, 255).astype(np.uint8)

    plt.figure(figsize=(12, 3))

    plt.subplot(1, 3, 1)
    plt.title(f"Face\nGT: {true_label}")
    plt.imshow(img_rgb); plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.title(f"LIME cut-out\nPred: {pred_label} (p={conf:.2f})")
    plt.imshow(cutout_uint8); plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.title("LIME overlay")
    plt.imshow(overlay); plt.axis("off")

    plt.tight_layout()
    plt.show()

    if save:
      base = f"lime_idx-{idx}_gt-{true_label}_pred-{pred_label}_p-{conf:.2f}"

      cutout_path = os.path.join(LIME_OUT, base + "_cutout.png")
      overlay_path = os.path.join(LIME_OUT, base + "_overlay.png")

      cv2.imwrite(cutout_path, cv2.cvtColor(cutout_uint8, cv2.COLOR_RGB2BGR))
      cv2.imwrite(overlay_path, cv2.cvtColor(overlay, cv2.COLOR_RGB2BGR))

    return {"idx": idx, "true_label": true_label, "pred_label": pred_label, "conf": conf}
```

```python
from matplotlib.gridspec import GridSpec

def _lime_cutout_for_pred(
    img_rgb_uint8,
    pred_idx,
    n_segments=120,
    compactness=10,
    sigma=1,
    num_samples=1500,
    num_features=15,
    positive_only=True,
    hide_color=0,
):

    segments = slic(
        img_as_ubyte(img_rgb_uint8),
        n_segments=int(n_segments),
        compactness=float(compactness),
        sigma=float(sigma),
        start_label=0,
    )

    explainer = lime_image.LimeImageExplainer(random_state=0)

    explanation = explainer.explain_instance(
        img_rgb_uint8,
        classifier_fn=deepface_predict_proba_lime,
        top_labels=1,
        hide_color=int(hide_color),
        num_samples=int(num_samples),
        segmentation_fn=lambda _: segments,
    )

    cutout, _mask = explanation.get_image_and_mask(
        label=int(pred_idx),
        positive_only=bool(positive_only),
        num_features=int(num_features),
        hide_rest=True,
    )

    cutout_uint8 = np.clip(cutout, 0, 255).astype(np.uint8)
    return cutout_uint8
```

```python
def compare_gradcam_occlusion_lime(
    indices,
    alpha=0.45,
    patch=6,
    stride=3,
    figsize_per_row=3.0,
    # LIME controls
    lime_n_segments=120,
    lime_num_samples=1500,
    lime_num_features=15,
    lime_hide_color=127,
    lime_positive_only=True,
    # Save
    save_path="/content/deepface_protocolA_outputs/grids/deepface_gradcam_occlusion_lime_grid.png",
    dpi=220,
):

    indices = [int(i) for i in indices]
    n = len(indices)

    out_dir = os.path.dirname(save_path)
    os.makedirs(out_dir, exist_ok=True)

    fig = plt.figure(figsize=(19, figsize_per_row * n), dpi=dpi)

    gs = GridSpec(
        nrows=n,
        ncols=5,
        figure=fig,
        width_ratios=[1.15, 3, 3, 3, 3],
        wspace=0.06,
        hspace=0.22,
    )

    for r, idx in enumerate(indices):
        img_bgr = image_data[idx]
        img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
        true_label = df.loc[idx, "emotion_label"]

        # DeepFace model input tensor
        img_tensor = preprocess_for_emotion(img_bgr)

        # Grad-CAMlogits returned by make_gradcam_heatmap
        gc_map, logits = make_gradcam_heatmap(img_tensor)
        pred_idx = int(np.argmax(logits))
        pred_label = EMOTIONS[pred_idx]

        probs = np.exp(logits - np.max(logits))
        probs = probs / (np.sum(probs) + 1e-8)
        conf = float(probs[pred_idx])

        # overlays
        _, _, gc_overlay = overlay_gradcam(img_rgb, gc_map, alpha=alpha)

        # Occlusion
        occ_map_48, _, _ = occlusion_sensitivity_map_48(
            img_tensor,
            class_index=pred_idx,
            patch=patch,
            stride=stride,
            baseline="mean",
        )
        _, _, occ_overlay = overlay_gradcam(img_rgb, occ_map_48, alpha=alpha)

        # LIME
        lime_cutout = _lime_cutout_for_pred(
            img_rgb_uint8=img_rgb,
            pred_idx=pred_idx,
            n_segments=lime_n_segments,
            compactness=10,
            sigma=1,
            num_samples=lime_num_samples,
            num_features=lime_num_features,
            positive_only=lime_positive_only,
            hide_color=lime_hide_color,
        )

        ax = fig.add_subplot(gs[r, 0])
        ax.axis("off")
        ax.text(
            0.0, 0.5,
            f"idx = {idx}\nGT = {true_label}\nPred = {pred_label}\np = {conf:.2f}",
            fontsize=12,
            va="center",
            ha="left",
        )

        # Face
        ax = fig.add_subplot(gs[r, 1])
        ax.imshow(img_rgb)
        if r == 0:
          ax.set_title("Face", fontsize=14, pad=8)
        ax.axis("off")

        # Grad-CAM overlay
        ax = fig.add_subplot(gs[r, 2])
        ax.imshow(gc_overlay)
        if r == 0:
            ax.set_title("Grad-CAM", fontsize=14, pad=8)
        ax.axis("off")

        # Occlusion overlay
        ax = fig.add_subplot(gs[r, 3])
        ax.imshow(occ_overlay)
        if r == 0:
            ax.set_title("Occlusion", fontsize=14, pad=8)
        ax.axis("off")

        # LIME cut out
        ax = fig.add_subplot(gs[r, 4])
        ax.imshow(lime_cutout)
        if r == 0:
            ax.set_title("LIME", fontsize=14, pad=8)
        ax.axis("off")

    plt.tight_layout()
    fig.savefig(save_path, dpi=dpi)
    print("Saved:", save_path)

    plt.show()
```

## **Results: Comparative Explainability Analysis (DeepFace)**

Figure 5 presents a qualitative comparison of three explainability methods,**Grad-CAM**, **occlusion sensitivity**, and **LIME**,applied to the DeepFace emotion recognition model on selected RAF-DB test samples. For each image, we report the ground-truth label (GT), the model’s predicted emotion, and the associated confidence score. All explanations are generated with respect to the **same predicted class** to ensure a fair and consistent comparison across methods.

**Grad-CAM** produces smooth and spatially continuous heatmaps that highlight broad facial regions, most commonly around the eyes, cheeks, and mouth. These maps provide a global view of the model’s internal attention patterns, but in several cases they extend beyond clearly localized facial features, particularly in misclassified examples. This suggests that Grad-CAM captures general activation trends rather than strictly localized decision cues.

In contrast, **occlusion sensitivity** yields more structured and localized explanations. Regions corresponding to the eyes, nose, and mouth consistently show the highest sensitivity, indicating that occluding these areas leads to the largest drop in the predicted class score. Compared to Grad-CAM, occlusion maps offer a more direct indication of *causal importance*, as they explicitly measure how targeted perturbations affect the model’s output.

**LIME cut-out explanations** provide a complementary, model-agnostic perspective. LIME highlights sparse sets of superpixels that are locally sufficient to support the predicted class. In correctly classified samples (e.g., idx = 15), the selected regions largely overlap with salient facial features identified by Grad-CAM and occlusion. However, in several misclassified cases (e.g., idx = 0, idx = 10, idx = 20), LIME selects fragmented or partial facial regions, suggesting that the model’s decision may rely on limited or ambiguous visual evidence.

Overall, the comparison reveals both agreement and divergence across explanation methods. Grad-CAM captures **global attention trends**, occlusion sensitivity identifies **functionally critical facial regions**, and LIME exposes **sparse local evidence** driving individual predictions. Disagreements between these methods,especially in failure cases, highlight potential instability in the model’s decision-making process, which is particularly relevant for downstream bias, robustness, and ethical analysis.
